```python
df["Sales"] = df["Sales"].fillna(df["Sales"].median())
df
```

# Compute KPIs

```python
total_sales = df["Sales"].sum()
total_sales
```

```python
total_orders = df["Quantity Ordered"].nunique()
total_orders
```

```python
total_quantity = df["Price Each"].sum()
total_quantity
```

# Create Visuals

```python
df.groupby("Product")["Sales"].sum().plot(kind="bar", color=["lightBlue"])
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

Month_sales = df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()

Month_sales.plot(kind="line",marker="o")

plt.title("Month Sales")
plt.xlabel("Month")
plt.ylabel("Sales") 
plt.show()
```

```python
plt.hist(df["Sales"],bins=20)
plt.xlabel("Sales")
plt.ylabel("Frequency") 
plt.title("Sales Distribution") 
plt.show()
```
